```python
X=df.drop("Churn", axis=1)
y = df["Churn"]

```

```python

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
    
)
```

```python

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
numerical_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="median")),
    ("scaler",  RobustScaler())
])
numerical_pipeline
```

```python
categorical_pipeline=Pipeline([
    ('imputer',SimpleImputer(strategy='most_frequent')),
    ("encoder", OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])
categorical_pipeline
```

```python
preprocessor = ColumnTransformer([
    ('numeric', numerical_pipeline,numerical_columns),
    ('categoric',categorical_pipeline,categorical_columns)
])
preprocessor
```

```python
X_train_prepar = preprocessor.fit_transform(X_train)
```

```python
X_test_prepar =preprocessor.transform(X_test) 
```

```python
print('Train:', X_train_prepar.shape, '| Test:', X_test_prepar.shape)
```

### الـ Accuracy بتجاوب على سؤال واحد: 

كام توقع من كل 100 كان صح؟ مثلاً 0.80 يعني الموديل صح في 80% من العملاء.
المشكلة إن داتا الـ Churn imbalanced: أغلب العملاء بيفضلوا، وأقلية بس بتسيب. لو موديل قال "محدش هيسيب" للكل، هيجيب Accuracy حوالي 73-75% (نسبة العملاء اللي فضلوا)، وهو ما مسك ولا عميل واحد من اللي هيسيبوا.
وده بيظهر في جدولك: Logistic Regression وAdaBoost عندهم نفس الـ Accuracy (0.802)، بس الفرق بين 0.802 و0.75 (اللي ممكن يجيبه موديل "كسول") مش كبير زي ما الرقم بيوحي. عشان كده لازم نبص على باقي الـ metrics.

----------------------------------------------------------

### الـ Precision: لما الموديل يقول "هيسيب"، يبقى صح قد إيه؟

الـ Precision = من كل العملاء اللي الموديل قال عنهم "هيسيبوا"، كام فعلاً سابوا.
Logistic Regression عنده 0.649، يعني من كل 100 عميل بيحذر منهم، حوالي 65 فعلاً هيسيبوا و35 إنذار كاذب. الـ Precision مهمة لو تكلفة التواصل مع العميل عالية (خصم، مكالمة، عرض)، لأنك مش عايز تصرف على ناس مش هيمشوا.

----------------------------------------------------------------------------------------

### الـ Recall: من العملاء اللي فعلاً هيسيبوا، مسكنا كام؟
الـ Recall = من كل العملاء اللي فعلاً سابوا، كام الموديل اكتشفهم.
Logistic Regression عنده 0.553، يعني بيمسك حوالي 55% بس من اللي هيسيبوا، والـ 45% الباقيين بيعدوا من غير ما حد ياخد باله. دي الـ False Negatives، وفي مشكلة الـ Churn هي الأغلى: خسارة عميل أكتر تكلفة من عرض لعميل كان هيفضل أصلاً.

ولاحظ KNN: الـ Recall بتاعه (0.553) نفس Logistic Regression، بس الـ Precision بتاعه ضعيف (0.535)، يعني بيمسك نفس العدد بس بإنذارات كاذبة أكتر.

--------------------------------------------------------------------------------------------------------
### الـ F1: الموازنة بين الاتنين
فيه تعارض بين Precision وRecall: لو خليت الموديل يحذر من أي حد، الـ Recall يزيد والـ Precision تقع، والعكس. الـ F1 هو المتوسط التوافقي بينهم، فبيكافئ الموديل المتوازن ويعاقب اللي بيضحي بواحد عشان التاني.
عشان كده رتبنا الجدول بالـ F1، وLogistic Regression جه الأول (0.597) لأنه الأوازن.

---------------------------------------------------------------------

### `الخلاصة`
الـ Accuracy بتخدع في الداتا imbalanced، وفرقها بين الموديلز الأول صغير.
الـ Recall هو الأهم للـ business هنا، لأننا عايزين نمسك أكبر عدد من العملاء قبل ما يسيبوا.
الـ F1 معيار الترتيب العادل لأنه بيجمع Precision وRecall.
Logistic Regression هو الأفضل كـ baseline (F1 = 0.597)، وAdaBoost قريب جداً منه.
